Case Folding


```python
# Teks awal
teks_asli = "Pemrosesan teks adalah tahapan penting dalam analisis data. Pada tahap ini, sistem komputer akan membersihkan dan mengubah teks mentah menjadi format yang lebih terstruktur. Proses tersebut melibatkan beberapa langkah seperti mengubah huruf menjadi kecil, memecah kalimat menjadi kata, menghapus kata-kata yang tidak memiliki makna penting, dan mengembalikan kata berimbuhan ke bentuk dasarnya. Tujuannya adalah agar algoritma dapat bekerja dengan lebih efisien dan akurat."

# --- TAHAP 1: CASE FOLDING ---
teks_case_folding = teks_asli.lower()

print("=== HASIL CASE FOLDING ===")
print(teks_case_folding)
```

Tahap Tokenisasi


```python
import re

# --- TAHAP 2: TOKENISASI ---
# Menghapus tanda baca terlebih dahulu agar token bersih
teks_bersih = re.sub(r'[^\w\s]', '', teks_case_folding)

# Memecah teks menjadi list kata
tokens = teks_bersih.split()

print("\n=== HASIL TOKENISASI ===")
print(tokens)
```

Tahap Stopword


```python
from Sastrawi.StopWordRemover.StopWordRemoverFactory import StopWordRemoverFactory

# --- TAHAP 3: STOP WORD REMOVAL ---
# Inisialisasi StopWordRemover
factory = StopWordRemoverFactory()
stopword_remover = factory.create_stop_word_remover()

# Menggabungkan token kembali menjadi kalimat untuk dihapus stopword-nya
kalimat_gabung = ' '.join(tokens)

# Menghapus stopword
teks_tanpa_stopword = stopword_remover.remove(kalimat_gabung)

# Memecah kembali menjadi list token
tokens_tanpa_stopword = teks_tanpa_stopword.split()

print("\n=== HASIL STOP WORD REMOVAL ===")
print(tokens_tanpa_stopword)
```

Tahap Stemming

```python
from Sastrawi.Stemmer.StemmerFactory import StemmerFactory

# --- TAHAP 4: STEMMING ---
# Inisialisasi Stemmer
factory_stemmer = StemmerFactory()
stemmer = factory_stemmer.create_stemmer()

# Melakukan stemming pada setiap kata dalam list
tokens_stemming = []
for kata in tokens_tanpa_stopword:
    kata_dasar = stemmer.stem(kata)
    tokens_stemming.append(kata_dasar)

print("\n=== HASIL STEMMING ===")
print(tokens_stemming)
```
